```python
import numpy as np

b = np.array([[1, 2, 3], [4, 5, 6]])   # 2D array (2 rows, 3 cols)
ar = np.arange(0, 10, 2)
print(b)
```

## 1️⃣ Indexing & Slicing

Format: `array[rows, columns]`

**Single element**: row 0, column 2

```python
print(b[0, 2])
```

**Entire column**: `:` before the comma means *all rows*

```python
print(b[:, 2])
```

**Entire row**: `:` after the comma means *all columns*

```python
print(b[1, :])
```

**Sub-array**: before the comma = rows, after the comma = columns

```python
print(b[0:2, 0:2])
```

## 2️⃣ Reshape

- Changes the shape (e.g. 1D → 2D) without changing the data or its order
- The total number of elements must stay the same
- It returns a new view, so the original `ar` is not changed

```python
print(ar)
reshaped = ar.reshape(5, 1)
print(reshaped)
print(ar)   # original unchanged
```

For `b` of shape (2, 3), valid reshapes are: (3, 2), (6, 1), (1, 6)

```python
print(b.reshape(3, 2))
print(b.reshape(1, 6))
```

## 3️⃣ Flatten vs Ravel

- `flatten()` converts any array into 1D (always returns a **copy**)
- `ravel()` also gives 1D, but returns a **view** when possible (faster, less memory)

```python
flat = b.flatten()
print(flat)
print(b)
```

```python
raveled = b.ravel()
print(raveled)
```

## 4️⃣ Array Arithmetic (element-wise)

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x ** 2)
```

## 5️⃣ Broadcasting

NumPy automatically stretches the smaller array to match the bigger one, with no loops needed.

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
vector = np.array([10, 20, 30])

print(matrix + vector)
```

## ✅ Quick Summary

| Concept | Code | Result |
|---|---|---|
| Single element | `b[0, 2]` | one value |
| Column | `b[:, 2]` | whole column |
| Row | `b[1, :]` | whole row |
| Sub-array | `b[0:2, 0:2]` | block of rows & cols |
| Reshape | `ar.reshape(5, 1)` | new shape, same data |
| Flatten | `b.flatten()` | 1D copy |
| Ravel | `b.ravel()` | 1D view |
| Broadcasting | `matrix + vector` | auto-stretched math |
