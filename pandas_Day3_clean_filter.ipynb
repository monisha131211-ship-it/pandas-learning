# Pandas Practice: Filter, Clean and Process Data

A small messy student dataset to practice what I learned:
`loc`, `iloc`, filtering, cleaning, adding columns and rows, groupby.

## 1. Create the messy data
This data has real-world problems:
- extra spaces and different spellings (inconsistent)
- `forty` instead of `40` (inappropriate)
- missing values (`None`)
- a duplicate row

```python
import pandas as pd
import numpy as np

data = {
    'name':   ['Ravi', ' meena', 'ASHA', 'Kiran', 'Ravi', 'Divya', 'Suresh', 'Latha'],
    'city':   ['chennai', 'Chennai', 'MADURAI', 'Coimbatore', 'chennai', 'Madurai', None, 'CHENNAI'],
    'marks':  [85, 92, 'forty', 70, 85, None, 66, 78],
    'course': ['Python', 'Python', 'SQL', 'Python', 'Python', 'SQL', 'SQL', 'Python']
}

df = pd.DataFrame(data)
df
```

## 2. Look at the data

```python
print(df.shape)       # (rows, columns)
print()
df.info()             # column types and missing values
```

```python
df.isnull().sum()     # count missing values in each column
```

## 3. Clean the data

**Step 1: Remove the duplicate row**

```python
df = df.drop_duplicates()
df
```

**Step 2: Fix inconsistent names and cities** (remove spaces, make the first letter capital)

```python
df['name'] = df['name'].str.strip().str.title()
df['city'] = df['city'].str.strip().str.title()
df
```

**Step 3: Fix inappropriate data** (`forty` becomes `40`, then make the column a number)

```python
df['marks'] = df['marks'].replace('forty', 40)
df['marks'] = pd.to_numeric(df['marks'])
df.dtypes
```

**Step 4: Handle missing values**
- missing city, fill with `Unknown`
- missing marks, fill with the average marks

```python
df['city'] = df['city'].fillna('Unknown')
df['marks'] = df['marks'].fillna(df['marks'].mean().round())
df
```

## 4. Select data with `loc` and `iloc`
- `loc` uses labels
- `iloc` uses position numbers

```python
df = df.reset_index(drop=True)

print('loc  :', df.loc[0, 'name'])
print('iloc :', df.iloc[0, 0])
print()
print('First 3 rows, name and marks:')
df.loc[0:2, ['name', 'marks']]
```

## 5. Filter the data

```python
# Students with marks above 80
df[df['marks'] > 80]
```

```python
# Students from Chennai
df[df['city'] == 'Chennai']
```

```python
# Python course AND marks above 75
df[(df['course'] == 'Python') & (df['marks'] > 75)]
```

## 6. Add a new column
Grade is based on marks.

```python
def get_grade(m):
    if m >= 85:
        return 'A'
    elif m >= 70:
        return 'B'
    else:
        return 'C'

df['grade'] = df['marks'].apply(get_grade)
df
```

## 7. Add a new row

```python
df.loc[len(df)] = ['Nisha', 'Chennai', 88, 'Python', 'A']
df
```

## 8. Process the data: sort and group

```python
# Highest marks first
df.sort_values('marks', ascending=False)
```

```python
# Average marks for each course
df.groupby('course')['marks'].mean().round(1)
```

```python
# Number of students in each city
df['city'].value_counts()
```

## 9. Save the clean data

```python
df.to_csv('clean_students.csv', index=False)
print('Saved: clean_students.csv')
```

## Summary
| Step | Function |
|---|---|
| Remove duplicates | `drop_duplicates()` |
| Fix text | `str.strip()`, `str.title()` |
| Fix wrong values | `replace()`, `pd.to_numeric()` |
| Missing values | `isnull()`, `fillna()` |
| Select data | `loc`, `iloc` |
| Filter | `df[df['marks'] > 80]` |
| Add column | `df['grade'] = ...` |
| Add row | `df.loc[len(df)] = [...]` |
| Group | `groupby()` |
