## 1. Data quality issues (concept)

**Inconsistent data** = same value written in different ways.

**Inappropriate data** = value in the wrong format (word instead of number).

```python
# Inconsistent: same city, 3 different spellings
cities = ['chennai', 'CHENNAI', 'Chennai']
print('Inconsistent example:', cities)

# Inappropriate: 'forty' is text, it should be the number 40
ages = [25, 'forty', 30]
print('Inappropriate example:', ages)
```

## 2. Import packages and check version

```python
import pandas as pd
import numpy as np

print('Pandas version:', pd.__version__)
```

## 3. Series (1D)
One column of data. The index (0, 1, 2...) comes automatically.

```python
colors = pd.Series(['Red', 'Blue', 'Green', 'Yellow'])
print('Series of Colors:')
print(colors)
print('\nType:', type(colors))
```

## 4. Custom index
We give our own labels instead of 0, 1, 2...

```python
price = pd.Series([50, 20, 100], index=['pen', 'book', 'bag'])
print(price)
```

## 5. Data filtering
Get only the values we want.

```python
# Get one value using its label
print('Price of book:', price['book'])

# Get only items with price more than 30
print('\nPrice more than 30:')
print(price[price > 30])
```

## 6. Series from a dictionary
Key = index, Value = data.

```python
marks = pd.Series({'Ravi': 85, 'Meena': 92, 'Asha': 78})
print(marks)
```

## 7. DataFrame from a dictionary
A table with rows and columns. Each key becomes a column name.

```python
data = {
    'name': ['Ravi', 'Meena', 'Asha'],
    'marks': [85, 92, 78]
}
df = pd.DataFrame(data)
df
```
